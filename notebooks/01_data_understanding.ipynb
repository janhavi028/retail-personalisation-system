```python
import pandas as pd

def analyze_column_summary(df, cols):
    summary_data = []
    
    for col in cols:
        vc = df[col].value_counts(normalize=True)
        top_5 = df[col].value_counts().head(5).to_dict()
        top_5_pct = (vc.head(5).sum() * 100).round(2)
        
        summary_data.append({
            "Column": col,
            "Unique Values": df[col].nunique(),
            "Top 5 Categories (Counts)": top_5,
            "Top 5 Concentration (%)": f"{top_5_pct}%"
        })
        
    return pd.DataFrame(summary_data)

target_cols = ["prod_name", "section_name", "department_name"]
result_df = analyze_column_summary(articles, target_cols)
result_df
```

### 3. Interaction Behaviour 

```python
customer_interactions = (
    transactions_train
    .groupby("customer_id")
    .agg(
        TotalTrans=("price", "count"),
        UniqueArticle=("article_id", "nunique")
    )
)
bins = [0, 1, 2, 5, 10, 20, 50, float("inf")]
labels = ["1", "2", "3-5", "6-10", "11-20", "21-50", "51+"]

customer_interactions["TransactionBucket"] = pd.cut(
    customer_interactions["TotalTrans"],
    bins=bins,
    labels=labels
)

customer_interactions["UniqueArticleBucket"] = pd.cut(
    customer_interactions["UniqueArticle"],
    bins=bins,
    labels=labels
)
transaction_distribution = (
    customer_interactions["TransactionBucket"]
    .value_counts(normalize=True, sort=False)
    .mul(100)
)

unique_article_distribution = (
    customer_interactions["UniqueArticleBucket"]
    .value_counts(normalize=True, sort=False)
    .mul(100)
)

print(transaction_distribution)
print(unique_article_distribution)
```

```python
repeat_interactions = (
    transactions_train
    .groupby(["customer_id", "article_id"])
    .agg(TotalTrans=("price", "count"))
    .reset_index()
)
repeat_interactions["TotalTrans"].value_counts(normalize=True).sort_index().head(15) * 100
```

So repeat purchasing of the exact same article exists, but it isn't dominant. Roughly:

100 - 87.13 = 12.87\%
of observed customer–article pairs have more than one transaction.

```python
customers_with_repeats = (
    repeat_interactions
    .groupby("customer_id")["TotalTrans"]
    .max()
)
customers_with_repeats.sort_values(ascending=False)
```

#### Interaction Sparsity

```python
total_customers = customers["customer_id"].nunique()
total_articles = articles["article_id"].nunique()

observed_combinations = (
    transactions_train[["customer_id", "article_id"]]
    .drop_duplicates()
    .shape[0]
)

total_combinations = total_customers * total_articles

density = observed_combinations / total_combinations
sparsity = 1 - density

print("Unique customers:", total_customers)
print("Unique articles:", total_articles)
print("Observed customer-article pairs:", observed_combinations)
print("Density:", density)
print("Sparsity:", sparsity)
```

So:

Observed: only about 0.019% of all possible customer–article pairs have a purchase.

Unobserved: about 99.981% have no recorded purchase.

That's an extremely sparse interaction matrix, which is very normal for retail recommenders.

But there's an important conceptual distinction:
99.981% unobserved does not mean customers dislike 99.981% of the products.

### 4. Cold start Customers

```python
missing_customers = customers[~customers["customer_id"].isin(transactions_train["customer_id"])]
missing_customers
```

```python
missing_customers["club_member_status"].value_counts()
```

9,699 customers have no transactions in the available training-period transaction data, of whom 9,308 are currently ACTIVE members.

Because you don't have membership start dates or transaction history from before the observation window, you cannot distinguish between genuinely new members and established members who simply had no transactions during this period.

I would now think of them as “customers with no observed interaction history” rather than cold-start customers. There are potentially several populations mixed together:
genuinely new members → true cold start
long-standing but inactive members → sparse/inactive-user problem
members whose relevant historical activity falls before your dataset → data-window limitation
possibly members whose activity isn't captured by this transaction table
The PRE-CREATE group is particularly interesting. Those 382 may plausibly represent customers at an earlier stage of membership, but we shouldn't infer exactly what that status means without knowing the business definition.

## Key Findings

### Dataset and catalogue
- The dataset contains approximately 1.37 million customers and 105,000 articles, with over 31 million transaction records.
- Article metadata contains several levels of product categorisation. Product names are highly granular, while attributes such as section and department provide broader product groupings that may be useful for candidate generation, ranking features, segmentation, and cold-start recommendations.
- Missing data is relatively limited in the transaction and article data, although some customer attributes and article descriptions contain missing values.

### Interaction behaviour
- The customer–article interaction space is extremely sparse. Only approximately 0.019% of all possible customer–article combinations are observed, corresponding to approximately 99.981% sparsity.
- An unobserved interaction should not be interpreted as a negative preference. A customer may simply never have encountered or had an opportunity to purchase an article, making this an implicit-feedback recommendation problem.
- Customer interaction histories vary substantially, with a meaningful proportion of customers having relatively rich purchase histories.
- Approximately 87% of observed customer–article pairs contain a single transaction, while around 13% contain repeat purchases. Repeat behaviour therefore contains potentially useful preference information, although most customer–article relationships occur only once.
- 9,699 customers in the customer table have no observed transactions in the available training period. These cannot confidently be classified as new customers because customer join dates and earlier transaction history are unavailable. Recommendation strategies that depend on historical interactions will therefore require fallback logic for customers without observable history.

### Modelling implications
- The high sparsity and implicit nature of the interactions make collaborative-filtering approaches such as implicit ALS appropriate candidates for evaluation.
- Interaction strength requires experimentation: binary purchase signals and frequency-based signals are both reasonable approaches given the observed repeat-purchase behaviour.
- Customers with little or no interaction history will require alternative recommendation strategies, such as popularity, segment-based or content-based recommendations.
- Evaluation must use a temporal split rather than a random transaction split because the business objective is to recommend articles that customers are likely to purchase in a future period.
