```python
import pandas as pd

def analyze_column_summary(df, cols):
    summary_data = []
    
    for col in cols:
        vc = df[col].value_counts(normalize=True)
        top_5 = df[col].value_counts().head(5).to_dict()
        top_5_pct = (vc.head(5).sum() * 100).round(2)
        
        summary_data.append({
            "Column": col,
            "Unique Values": df[col].nunique(),
            "Top 5 Categories (Counts)": top_5,
            "Top 5 Concentration (%)": f"{top_5_pct}%"
        })
        
    return pd.DataFrame(summary_data)

target_cols = ["prod_name", "section_name", "department_name"]
result_df = analyze_column_summary(articles, target_cols)
result_df
```

### 3. Interaction Behaviour 

```python
customer_interactions = (
    transactions_train
    .groupby("customer_id")
    .agg(
        TotalTrans=("price", "count"),
        UniqueArticle=("article_id", "nunique")
    )
)
bins = [0, 1, 2, 5, 10, 20, 50, float("inf")]
labels = ["1", "2", "3-5", "6-10", "11-20", "21-50", "51+"]

customer_interactions["TransactionBucket"] = pd.cut(
    customer_interactions["TotalTrans"],
    bins=bins,
    labels=labels
)

customer_interactions["UniqueArticleBucket"] = pd.cut(
    customer_interactions["UniqueArticle"],
    bins=bins,
    labels=labels
)
transaction_distribution = (
    customer_interactions["TransactionBucket"]
    .value_counts(normalize=True, sort=False)
    .mul(100)
)

unique_article_distribution = (
    customer_interactions["UniqueArticleBucket"]
    .value_counts(normalize=True, sort=False)
    .mul(100)
)

print(transaction_distribution)
print(unique_article_distribution)
```

```python
repeat_interactions = (
    transactions_train
    .groupby(["customer_id", "article_id"])
    .agg(TotalTrans=("price", "count"))
    .reset_index()
)
repeat_interactions["TotalTrans"].value_counts(normalize=True).sort_index().head(15) * 100
```

So repeat purchasing of the exact same article exists, but it isn't dominant. Roughly:

100 - 87.13 = 12.87\%
of observed customer–article pairs have more than one transaction.

```python
customers_with_repeats = (
    repeat_interactions
    .groupby("customer_id")["TotalTrans"]
    .max()
)
customers_with_repeats.sort_values(ascending=False)
```

#### Interaction Sparsity

```python
total_customers = customers["customer_id"].nunique()
total_articles = articles["article_id"].nunique()

observed_combinations = (
    transactions_train[["customer_id", "article_id"]]
    .drop_duplicates()
    .shape[0]
)

total_combinations = total_customers * total_articles

density = observed_combinations / total_combinations
sparsity = 1 - density

print("Unique customers:", total_customers)
print("Unique articles:", total_articles)
print("Observed customer-article pairs:", observed_combinations)
print("Density:", density)
print("Sparsity:", sparsity)
```

So:

Observed: only about 0.019% of all possible customer–article pairs have a purchase.

Unobserved: about 99.981% have no recorded purchase.

That's an extremely sparse interaction matrix, which is very normal for retail recommenders.

But there's an important conceptual distinction:
99.981% unobserved does not mean customers dislike 99.981% of the products.
